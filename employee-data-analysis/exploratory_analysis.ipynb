```python
categorical_columns = df.select_dtypes(
    include=["str", "object"]
).columns

for col in categorical_columns:
    print(f"\n{col}:")
    print(df[col].unique())
```

```python
clean_df = df.copy()
```

```python
for col in categorical_columns:
    clean_df[col] = clean_df[col].str.strip()
```

```python
for col in categorical_columns:
    print(col, clean_df[col].unique())
```

```python
rows_before = len(clean_df)

duplicates_before = clean_df.duplicated().sum()

clean_df = clean_df.drop_duplicates()

rows_after = len(clean_df)

print("Rows before:", rows_before)
print("Duplicates found:", duplicates_before)
print("Rows after:", rows_after)
```

```python
numeric_columns = clean_df.select_dtypes(
    include="number"
).columns

exclude_columns = [
    "EmployeeNumber",
    "EmployeeCount",
    "StandardHours"
]

outlier_columns = [
    col for col in numeric_columns
    if col not in exclude_columns
]

print(outlier_columns)
```

```python
outlier_summary = []

for col in outlier_columns:

    Q1 = clean_df[col].quantile(0.25)
    Q3 = clean_df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = clean_df[
        (clean_df[col] < lower_limit) |
        (clean_df[col] > upper_limit)
    ]

    outlier_summary.append({
        "Column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Limit": lower_limit,
        "Upper Limit": upper_limit,
        "Outlier Count": len(outliers)
    })

outlier_df = pd.DataFrame(outlier_summary)

print(outlier_df.to_string(index=False))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.boxplot(x=clean_df["MonthlyIncome"])

plt.title("Monthly Income - Outlier Detection")
plt.xlabel("Monthly Income")

plt.show()
```

```python
before_after_summary = pd.DataFrame({
    "Metric": ["Rows", "Missing Values", "Duplicates"],
    "Before": [
        len(df),
        df.isnull().sum().sum(),
        df.duplicated().sum()
    ],
    "After": [
        len(clean_df),
        clean_df.isnull().sum().sum(),
        clean_df.duplicated().sum()
    ]
})

print(before_after_summary)
```

```python
clean_df.to_csv("data/clean_data.csv", index=False)

before_after_summary.to_csv(
    "data/before_after_summary.csv",
    index=False
)

print("Files saved successfully!")
```
