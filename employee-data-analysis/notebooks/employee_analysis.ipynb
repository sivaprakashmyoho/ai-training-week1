```python
df = pd.read_csv(
    "data/employees.csv",
    encoding="utf-16",
    header=None,
    names=["ID", "Name", "City"]
)

print(df)
```

```python
print(df.shape)
df.info()
```

```python
print("Total customers:", len(df))
```

```python
print(df["City"].value_counts())
```

```python
print("Chennai customers:", (df["City"] == "Chennai").sum())
```

```python
print("Unique cities:", df["City"].unique())
```

```python
print(df["City"].value_counts(normalize=True) * 100)
```

```python

# Customer Data Analysis - Project Summary

print("CUSTOMER DATA ANALYSIS")
print("----------------------")

# 1. Total customers
print("\n1. Total Customers:", len(df))

# 2. Customers by city
print("\n2. Customers by City:")
print(df["City"].value_counts())

# 3. Chennai customers
print("\n3. Chennai Customers:",
      (df["City"] == "Chennai").sum())

# 4. Unique cities
print("\n4. Unique Cities:")
print(df["City"].unique())

# 5. Customer percentage by city
print("\n5. Customer Percentage by City:")
print(df["City"].value_counts(normalize=True) * 100)

# Final data quality summary
print("\nDATA QUALITY")
print("------------")
print("Missing Values:", df.isna().sum().sum())
print("Duplicate Rows:", df.duplicated().sum())
```
